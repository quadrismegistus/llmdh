```python
import pandas as pd
import os
from tqdm.auto import tqdm
```

```python
folder = '/Users/ryan/GDrive/Bechdel/Data/parsed_movie_scripts'
df = pd.read_pickle('./data.bechdel.dfall.pkl')
o=[]
for fn in tqdm(os.listdir(folder)):
    if not fn.endswith('.csv'): continue
    fnid = os.path.splitext(fn)[0]
    if fnid not in df.index: continue
    fnfn = os.path.join(folder, fn)
    fndf = pd.read_csv(fnfn).fillna('')
    fn_o=[]
    for scene_num, scenedf in fndf.groupby('scene_num'):
        if len(scenedf)<10: continue
        txts = [scenedf.iloc[0].scene_desc]
        for i,row in scenedf.iterrows():
            if row.narration: txts.append(f'{row.narration}')
            speech = [f'\t{row.speaker.upper()}:']
            if row.direction: speech.append(f'({row.direction})')
            speech.append(row.speech)
            txts.append(' '.join(speech))
        txt = '\n\n'.join(txts)

        odx = {
            'id':fnid,
            'scene_num':scene_num,
            'scene_txt':txt,
            'scene_num_words': len(txt.split())
        }
        if 100 <= odx['scene_num_words'] <= 1000:
            fn_o.append(odx)
    if len(fn_o)>=5:
        o.extend(fn_o)
dftxt = pd.DataFrame(o)
odf = dftxt.merge(df[['title', 'source', 'release_date', 'overview', 'imdb_id']], on='id', how='left')
odf['release_date'] = odf['release_date'].fillna(0).astype(int)
odf['imdb_id'] = odf['imdb_id'].fillna(0).astype(int).astype(str)
odf
```
