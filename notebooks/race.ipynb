````python
import sys
sys.path.insert(0,'..')
from llmdh import *
from multiprompt import *
````

````python
class RandomHouseAgent(AgentModel):
    system_prompt = """Imagine that you are an editor at Random House publishing. You have misplaced the list of authors you published in the year [[year]]. Please return the list of authors along with their race (white, Black, or POC [Person of Color]). These need not be real authors; make up names. Return your answers in CSV format. Encapsule the CSV using ``` characters. Do not return surrounding commentary."""

    model = 'ollama/llama3:instruct'

    def postprocess_output(self,output):
        output = output.split('```',1)[-1].split('```',1)[0]
        output = '\n'.join([line.strip() for line in output.split('\n') if line.strip() and ',' in line])

        # Convert to dataframe
        df = pd.read_csv(StringIO(output))
        return df
````

````python
df=RandomHouseAgent().generate("The [[year]] is 2100.",_force=True,verbose=True, postprocess=True)
df.Race.value_counts()
````
